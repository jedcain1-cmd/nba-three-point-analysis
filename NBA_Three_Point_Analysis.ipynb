```python
all_games = []

for team in teams:
    # Load both files
    adv = pd.read_csv(f"Advanced_Game_Logs/{team}_2026_Advanced.csv")
    basic = pd.read_csv(f"Basic_Game_Logs/{team}_2026_Basic.csv")

    # Remove the season-total row from the Basic log
    basic = basic.dropna(subset=['Gtm'])

    # Keep only the Basic columns we need
    basic = basic[['Gtm', '3P', '3PA']]

    # Merge Basic shooting data into Advanced data by game number
    merged = adv.merge(basic, on='Gtm', how='inner')

    # Add team abbreviation
    merged['Team'] = team

    # Calculate game-level 3P%
    merged['3P%_Game'] = merged['3P'] / merged['3PA']

    # Add this team's games to our master dataset
    all_games.append(merged)

# Combine all 30 teams
nba_games = pd.concat(all_games, ignore_index=True)

nba_games
```

```python
import pandas as pd
import time
from nba_api.stats.endpoints import leaguegamelog, teamgamelogs

seasons = [
    '2016-17', '2017-18', '2018-19', '2019-20', '2020-21',
    '2021-22', '2022-23', '2023-24', '2024-25', '2025-26'
]

all_playoffs = []

for season in seasons:
    print(f"Downloading {season} playoffs...")

    # Traditional stats
    trad = leaguegamelog.LeagueGameLog(
        season=season,
        season_type_all_star='Playoffs',
        player_or_team_abbreviation='T'
    ).get_data_frames()[0]

    time.sleep(1)

    # Advanced stats
    adv = teamgamelogs.TeamGameLogs(
        season_nullable=season,
        season_type_nullable='Playoffs',
        measure_type_player_game_logs_nullable='Advanced'
    ).get_data_frames()[0]

    # Keep advanced columns we need
    adv = adv[
        ['GAME_ID', 'TEAM_ID', 'OFF_RATING', 'DEF_RATING', 'PACE']
    ]

    # Merge
    merged = trad.merge(
        adv,
        on=['GAME_ID', 'TEAM_ID'],
        how='left'
    )

    merged['SEASON'] = season
    all_playoffs.append(merged)

    print(f"  {len(merged)} team-game rows")
    time.sleep(1)

playoffs_all = pd.concat(all_playoffs, ignore_index=True)

print("\nFinished!")
print("Total rows:", len(playoffs_all))
print("Missing OFF_RATING:", playoffs_all['OFF_RATING'].isna().sum())
```

```python
playoffs_all.to_csv('NBA_Playoffs_2017_2026.csv', index=False)

print("Saved!")
```

```python
regular_seasons = []

for season in seasons:
    print(f"Downloading {season} regular season...")

    # Traditional stats
    trad = leaguegamelog.LeagueGameLog(
        season=season,
        season_type_all_star='Regular Season',
        player_or_team_abbreviation='T'
    ).get_data_frames()[0]

    time.sleep(1)

    # Advanced stats
    adv = teamgamelogs.TeamGameLogs(
        season_nullable=season,
        season_type_nullable='Regular Season',
        measure_type_player_game_logs_nullable='Advanced'
    ).get_data_frames()[0]

    adv = adv[
        ['GAME_ID', 'TEAM_ID', 'OFF_RATING', 'DEF_RATING', 'PACE']
    ]

    merged = trad.merge(
        adv,
        on=['GAME_ID', 'TEAM_ID'],
        how='left'
    )

    merged['SEASON'] = season
    regular_seasons.append(merged)

    print(f"  {len(merged)} team-game rows")
    time.sleep(1)

regular_all = pd.concat(regular_seasons, ignore_index=True)

print("\nFinished!")
print("Total rows:", len(regular_all))
print("Missing OFF_RATING:", regular_all['OFF_RATING'].isna().sum())
```

```python
regular_all['Win'] = (regular_all['WL'] == 'W').astype(int)
regular_all['3PAr'] = regular_all['FG3A'] / regular_all['FGA']

regular_baselines = (
    regular_all
    .groupby(['SEASON', 'TEAM_ID', 'TEAM_ABBREVIATION'])
    .agg(
        Reg_3PAr=('3PAr', 'mean'),
        Reg_3P_Pct=('FG3_PCT', 'mean'),
        Reg_ORtg=('OFF_RATING', 'mean'),
        Reg_Win_Pct=('Win', 'mean'),
        Reg_DRtg=('DEF_RATING', 'mean'),
        Reg_Pace=('PACE', 'mean')
    )
    .reset_index()
)

print(regular_baselines.shape)
regular_baselines.head()
```

```python
playoffs_with_baseline = playoffs_all.merge(
    regular_baselines,
    on=['SEASON', 'TEAM_ID', 'TEAM_ABBREVIATION'],
    how='left'
)

print(playoffs_with_baseline.shape)
print(playoffs_with_baseline[
    ['SEASON', 'TEAM_ABBREVIATION', 'FG3_PCT', 'OFF_RATING',
     'Reg_3PAr', 'Reg_3P_Pct', 'Reg_ORtg']
].head())

print("Missing baseline rows:",
      playoffs_with_baseline['Reg_ORtg'].isna().sum())
```

```python
playoffs_with_baseline['3P_Pct_Dev'] = (
    playoffs_with_baseline['FG3_PCT']
    - playoffs_with_baseline['Reg_3P_Pct']
)

playoffs_with_baseline['ORtg_Dev'] = (
    playoffs_with_baseline['OFF_RATING']
    - playoffs_with_baseline['Reg_ORtg']
)

playoffs_with_baseline['Reliance_x_3P_Dev'] = (
    playoffs_with_baseline['Reg_3PAr']
    * playoffs_with_baseline['3P_Pct_Dev']
)

print(
    playoffs_with_baseline[
        ['SEASON', 'TEAM_ABBREVIATION',
         'Reg_3PAr', '3P_Pct_Dev', 'ORtg_Dev']
    ].head(10)
)
```

```python
import statsmodels.formula.api as smf

playoff_model = smf.ols(
    formula="""
        ORtg_Dev ~ 3P_Pct_Dev
        + Reg_3PAr
        + 3P_Pct_Dev:Reg_3PAr
        + C(SEASON)
        + C(TEAM_ABBREVIATION)
    """,
    data=playoffs_with_baseline
).fit(
    cov_type='cluster',
    cov_kwds={'groups': playoffs_with_baseline['GAME_ID']}
)

print(playoff_model.summary())
```

```python
import numpy as np
import pandas as pd

np.random.seed(42)

n_games = 100000
fga = 88

two_pct = overall_2p_pct
three_pct = overall_3p_pct

three_rates = [0.30, 0.35, 0.40, 0.45, 0.50, 0.55]

simulation_results = []

for rate in three_rates:
    
    three_attempts = round(fga * rate)
    two_attempts = fga - three_attempts
    
    three_makes = np.random.binomial(
        three_attempts,
        three_pct,
        n_games
    )
    
    two_makes = np.random.binomial(
        two_attempts,
        two_pct,
        n_games
    )
    
    points = 3 * three_makes + 2 * two_makes
    
    simulation_results.append({
        '3PA_Rate': rate,
        '3PA': three_attempts,
        '2PA': two_attempts,
        'Avg_Points': points.mean(),
        'Points_SD': points.std(),
        '10th_Percentile': np.percentile(points, 10),
        '90th_Percentile': np.percentile(points, 90)
    })

simulation_results = pd.DataFrame(simulation_results)

print(simulation_results)
```
